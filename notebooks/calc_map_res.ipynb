```python
import gc
import numpy as np
from ggner_3d.cm import CLR_CONNS


def calculate_map_resolution():
    resolutions_to_check = [10_000]

    threshold_frac = 0.8
    min_contacts = 1000

    passing_resolutions = {}

    print(f"Available resolutions: {resolutions_to_check}", flush=True)

    for res in resolutions_to_check:
        clrs_ = CLR_CONNS(res)

        for sample_name, clr in clrs_.items():
            print(f"\nProcessing {sample_name} @ {res} bp", flush=True)

            mat = clr.matrix(balance=False, sparse=True)[:]

            coverage = np.asarray(mat.sum(axis=0)).ravel()

            total_bins = coverage.size
            high_cov_bins = np.count_nonzero(coverage >= min_contacts)

            print(
                f"Resolution: {res} bp - "
                f"Bins ≥ {min_contacts}: {high_cov_bins}/{total_bins} "
                f"({high_cov_bins / total_bins * 100:.2f}%)",
                flush=True
            )

            if high_cov_bins >= threshold_frac * total_bins:
                passing_resolutions.setdefault(sample_name, []).append(res)
            else:
                print(f"Failed at {res} bp for {sample_name}", flush=True)
                break

            del mat
            del coverage
            gc.collect()

        del clrs_
        gc.collect()

    return passing_resolutions


if __name__ == "__main__":
    res = calculate_map_resolution()
    print("\nPassing resolutions:")
    print(res)
```
